# MNIST

```python
import numpy as np
from helper_functions import *
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
```

```python
X_train = load_images('mnist/train-images-idx3-ubyte/train-images-idx3-ubyte')
y_train = load_labels('mnist/train-labels-idx1-ubyte/train-labels-idx1-ubyte')

X_test = load_images('mnist/t10k-images-idx3-ubyte/t10k-images-idx3-ubyte')
y_test = load_labels('mnist/t10k-labels-idx1-ubyte/t10k-labels-idx1-ubyte')

y_train = y_train.astype(int)
y_test = y_test.astype(int)

# Sort training set
train_idx = np.argsort(y_train)
X_train = X_train[train_idx]
y_train = y_train[train_idx]

# Sort test set
test_idx = np.argsort(y_test)
X_test = X_test[test_idx]
y_test = y_test[test_idx]
```

```python
plt.imshow(X_train[5999], cmap='gray')
plt.title(y_train[6000])
plt.axis('off')
plt.show()
```

## Preprocessing

```python
X_train = X_train.reshape(-1,28*28).T
X_test = X_test.reshape(-1,28*28).T
N_train = X_train.shape[0]
N_test = X_test.shape[0]

#PCA
A,mean = PCA(X_train,64)
X_train_PCA = A@(X_train - mean)
X_test_PCA = A@(X_test - mean)
```

```python
i=0
img =A[i,:].reshape(28,28)
plt.imshow(img, cmap='gray')
plt.title(y_train[i])
plt.axis('off')
plt.show()
```

# SVM

```python
%%time
Xtr = X_train_PCA.T
Xte = X_test_PCA.T
ytr = y_train
yte = y_test
sc = StandardScaler()
Xtr = sc.fit_transform(Xtr)
Xte = sc.transform(Xte)
clf = SVC(kernel='linear')
clf.fit(Xtr, ytr)

pred = clf.predict(Xte)
print("Linear testing accuracy:", accuracy_score(yte, pred))
train_pred = clf.predict(Xtr)
print("Linear Training accuracy:", accuracy_score(ytr, train_pred))
```

```python
%%time
# Polynomial kernel
poly_clf = SVC(kernel='poly', degree=3, C=1.0, gamma='scale')
poly_clf.fit(Xtr, ytr)


pred = poly_clf.predict(Xte)
print("Polynomial testing accuracy:", accuracy_score(yte, pred))
train_pred = poly_clf.predict(Xtr)
print("Polynomial Training accuracy:", accuracy_score(ytr, train_pred))
```

```python
%%time
# RBF kernel
rbf_clf = SVC(kernel='rbf', C=1.0, gamma='scale')
rbf_clf.fit(Xtr, ytr)


pred = rbf_clf.predict(Xte)
print("RBF testing accuracy:", accuracy_score(yte, pred))
train_pred = rbf_clf.predict(Xtr)
print("RBF Training accuracy:", accuracy_score(ytr, train_pred))
```

### LDA to 9 dimensions

```python
# LDA with 9 dimensions
lda = LinearDiscriminantAnalysis(n_components=9)

Xtr_lda = lda.fit_transform(Xtr, ytr)
Xte_lda = lda.transform(Xte)

print("LDA train shape:", Xtr_lda.shape)
print("LDA test shape:", Xte_lda.shape)

clf = SVC(kernel='linear')
clf.fit(Xtr_lda, ytr)

pred = clf.predict(Xte_lda)
print("Linear testing accuracy:", accuracy_score(yte, pred))
train_pred = clf.predict(Xtr_lda)
print("Linear Training accuracy:", accuracy_score(ytr, train_pred))
```

```python
# Polynomial kernel
poly_clf = SVC(kernel='poly', degree=3, C=1.0, gamma='scale')
poly_clf.fit(Xtr_lda, ytr)


pred = poly_clf.predict(Xte_lda)
print("Polynomial testing accuracy:", accuracy_score(yte, pred))
train_pred = poly_clf.predict(Xtr_lda)
print("Polynomial Training accuracy:", accuracy_score(ytr, train_pred))
```

```python
%%time
# RBF kernel
rbf_clf = SVC(kernel='rbf', C=1.0, gamma='scale')
rbf_clf.fit(Xtr_lda, ytr)


pred = rbf_clf.predict(Xte_lda)
print("RBF testing accuracy:", accuracy_score(yte, pred))
train_pred = rbf_clf.predict(Xtr_lda)
print("RBF Training accuracy:", accuracy_score(ytr, train_pred))
```
